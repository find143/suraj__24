# 🛒 Walmart Sales Forecasting Project

This notebook demonstrates a complete time-series forecasting pipeline to predict future sales based on historical sales data.

## 📋 Project Steps:
1. **Data Prep & Simulation**: Set up synthetic sales data mirroring the structure of Walmart sales.
2. **Feature Engineering**: Create time-based features (day, month, lag values, rolling averages).
3. **Exploratory Data Analysis (EDA)**: Plot sales trends and perform seasonal decomposition.
4. **Model Development**: Train regression models (Random Forest and XGBoost) with time-aware train/test splitting.
5. **Model Evaluation & Visualizations**: Compare actual vs. predicted values over time.

````python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

# Set plotting styles
plt.style.use('fivethirtyeight')
sns.set_theme(style="whitegrid")
````

## 📊 Step 1: Data Generation / Loading

We will programmatically generate a realistic daily sales dataset representing multiple Walmart departments/stores to ensure our forecasting models can run and be evaluated immediately.

````python
# Generate 3 years of daily sales data for 2 different stores to make it realistic
np.random.seed(42)
dates = pd.date_range(start="2021-01-01", end="2023-12-31", freq="D")
n_days = len(dates)

data_list = []
for store in [1, 2]:
    # Baseline sales with weekly seasonality, yearly trend, and random noise
    base_sales = 20000 if store == 1 else 15000

    # Day of week effect (higher sales on weekends)
    day_effect = [1.0, 0.95, 0.98, 1.02, 1.15, 1.35, 1.25]

    # Month effect (peak holiday season in Nov/Dec)
    month_effect = {1: 0.9, 2: 0.95, 3: 1.0, 4: 1.0, 5: 1.05, 6: 1.1,
                    7: 1.1, 8: 1.05, 9: 1.0, 10: 1.05, 11: 1.3, 12: 1.5}

    for idx, d in enumerate(dates):
        dow = d.weekday()
        mon = d.month
        # Add a light upward trend over the years
        trend = 1 + (idx / n_days) * 0.15
        noise = np.random.normal(0, 0.05)

        weekly_factor = day_effect[dow]
        monthly_factor = month_effect[mon]

        sales = base_sales * trend * weekly_factor * monthly_factor * (1 + noise)

        # Mark specific holidays
        is_holiday = int(d.month == 11 and d.day >= 24) or int(d.month == 12 and d.day >= 20)
        if is_holiday:
            sales *= 1.25 # Holiday boost

        data_list.append({
            'Date': d,
            'Store': store,
            'Weekly_Sales': round(sales, 2),
            'IsHoliday': bool(is_holiday)
        })

df = pd.DataFrame(data_list)
display(df.head(10))
print(f"Dataset shape: {df.shape}")
````

## ⚙️ Step 2: Feature Engineering

To help our regression models forecast next-period sales accurately, we need to build strong predictive features:
1. **Time-based Features**: `Day`, `Month`, `Year`, `DayOfWeek` (extracted from the `Date` index).
2. **Lag Features**: Past sales (e.g., sales from 1 day ago, 7 days ago) to capture recent patterns.
3. **Rolling Statistics**: Rolling average over the past 7 days to smooth out random noise and reveal underlying trends.

````python
# Sort dataset to ensure time sequences are correct before computing lags
df = df.sort_values(by=['Store', 'Date']).reset_index(drop=True)

# 1. Time-based features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek

# 2. Lag features (grouped by store to avoid leaking sales data between stores)
df['Sales_Lag_1'] = df.groupby('Store')['Weekly_Sales'].shift(1)
df['Sales_Lag_7'] = df.groupby('Store')['Weekly_Sales'].shift(7)

# 3. Rolling window features
df['Sales_Rolling_Mean_7'] = df.groupby('Store')['Weekly_Sales'].transform(lambda x: x.shift(1).rolling(window=7).mean())

# Drop rows with NaN values resulting from lags and rolling windows
df_clean = df.dropna().reset_index(drop=True)

print(f"Dataset shape after removing NaN values from feature engineering: {df_clean.shape}")
display(df_clean.head(5))
````

## 📈 Step 3: Exploratory Data Analysis & Seasonal Decomposition

Let's visualize the engineered features and look at the overall trends, weekly fluctuations, and seasonality.

````python
# Plot daily sales for Store 1 and Store 2 to see trends and seasonal peaks
plt.figure(figsize=(15, 6))
for store in df_clean['Store'].unique():
    store_data = df_clean[df_clean['Store'] == store]
    plt.plot(store_data['Date'], store_data['Weekly_Sales'], label=f'Store {store}', alpha=0.8)

plt.title('Daily Sales Trend Over Time (Store 1 vs Store 2)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Sales ($)')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

# Visualize weekly seasonality and Holiday effects with clean Seaborn configurations
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Average sales by day of week
sns.barplot(data=df_clean, x='DayOfWeek', y='Weekly_Sales', ax=axes[0], hue='DayOfWeek', palette='Blues_d', legend=False)
axes[0].set_title('Average Sales by Day of Week (0=Monday, 6=Sunday)')
axes[0].set_xticks(range(7))
axes[0].set_xticklabels(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])

# Holiday vs Non-Holiday impact
sns.boxplot(data=df_clean, x='IsHoliday', y='Weekly_Sales', ax=axes[1], hue='IsHoliday', palette='Set2', legend=False)
axes[1].set_title('Sales Distribution: Holiday vs Regular Days')

plt.tight_layout()
plt.show()
````

## 🤖 Step 4: Model Development with Time-aware Validation

For time-series data, standard random train/test splits are inappropriate because they cause data leakage (predicting the past using future information). Instead, we will use a **Time-Series Split** (splitting chronologically).

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
try:
    from xgboost import XGBRegressor
except ImportError:
    # Fallback to importing from sklearn if xgboost is unavailable in the environment
    from sklearn.ensemble import GradientBoostingRegressor as XGBRegressor

# Sort chronologically
df_clean = df_clean.sort_values(by='Date').reset_index(drop=True)

# Define cutoff date for 80/20 chronological split
split_idx = int(len(df_clean) * 0.8)
split_date = df_clean.iloc[split_idx]['Date']

train = df_clean[df_clean['Date'] < split_date]
test = df_clean[df_clean['Date'] >= split_date]

# Define features and target variable
features = ['Store', 'IsHoliday', 'Year', 'Month', 'Day', 'DayOfWeek', 'Sales_Lag_1', 'Sales_Lag_7', 'Sales_Rolling_Mean_7']
target = 'Weekly_Sales'

X_train, y_train = train[features], train[target]
X_test, y_test = test[features], test[target]

print(f"Training set date range: {train['Date'].min().strftime('%Y-%m-%d')} to {train['Date'].max().strftime('%Y-%m-%d')}")
print(f"Testing set date range:  {test['Date'].min().strftime('%Y-%m-%d')} to {test['Date'].max().strftime('%Y-%m-%d')}")
````

````python
# Initialize models
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
xgb_model = XGBRegressor(n_estimators=100, learning_rate=0.05, random_state=42)

# Fit models
print("Training Random Forest...")
rf_model.fit(X_train, y_train)

print("Training XGBoost...")
xgb_model.fit(X_train, y_train)

# Predict
rf_preds = rf_model.predict(X_test)
xgb_preds = xgb_model.predict(X_test)
print("Modeling complete!")
````

## 📊 Step 5: Model Evaluation & Visualization

Let's calculate metrics (MAE, RMSE, R²) and visualize how well our models' forecasts match the actual sales over time.

````python
def evaluate_forecast(y_true, y_pred, model_name):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"=== {model_name} Evaluation ===")
    print(f"Mean Absolute Error (MAE):     ${mae:.2f}")
    print(f"Root Mean Squared Error (RMSE): ${rmse:.2f}")
    print(f"R-squared Score (R2):          {r2:.4f}\n")
    return mae, rmse, r2

rf_metrics = evaluate_forecast(y_test, rf_preds, "Random Forest")
xgb_metrics = evaluate_forecast(y_test, xgb_preds, "XGBoost")
````

````python
# Plot actual vs predicted sales for the test period
test_results = test.copy()
test_results['RF_Predicted'] = rf_preds
test_results['XGB_Predicted'] = xgb_preds

# Filter to one store (Store 1) for a clean visual comparison
store_test = test_results[test_results['Store'] == 1].sort_values(by='Date')

plt.figure(figsize=(15, 7))
plt.plot(store_test['Date'], store_test['Weekly_Sales'], label='Actual Sales', color='black', alpha=0.6, linewidth=2)
plt.plot(store_test['Date'], store_test['RF_Predicted'], label='Random Forest Forecast', color='orange', linestyle='--', alpha=0.9)
plt.plot(store_test['Date'], store_test['XGB_Predicted'], label='XGBoost Forecast', color='green', linestyle=':', alpha=0.9)

plt.title('Store 1: Actual vs. Forecasted Sales (Test Period)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Sales ($)')
plt.legend()
plt.tight_layout()
plt.show()
````
